```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

sns.set_style("whitegrid")
```

```python
fraud = pd.read_csv("../data/raw/Fraud_Data.csv")

fraud.head()
```

```python
fraud.shape
```

```python
fraud.info()
```

```python
fraud.describe()
```

```python
fraud.isnull().sum()
```

```python
plt.figure(figsize=(12,5))

fraud.isnull().sum().sort_values().plot(kind="bar")

plt.title("Missing Values")
plt.show()
```

```python
fraud.duplicated().sum()
```

```python
fraud = fraud.drop_duplicates()
```

```python
fraud["signup_time"] = pd.to_datetime(
    fraud["signup_time"]
)

fraud["purchase_time"] = pd.to_datetime(
    fraud["purchase_time"]
)
```

```python
fraud["class"].value_counts()
```

```python
fraud["class"].value_counts(normalize=True)*100
```

```python
sns.countplot(
    x="class",
    data=fraud
)

plt.title("Fraud Class Distribution")
plt.show()
```

Dataset is highly imbalanced.
Fraudulent transactions represent a small minority.
Accuracy alone will be misleading.

```python
plt.figure(figsize=(10,5))

sns.histplot(
    fraud["purchase_value"],
    bins=50
)

plt.show()
```

```python
sns.boxplot(
    x="class",
    y="purchase_value",
    data=fraud
)
```
