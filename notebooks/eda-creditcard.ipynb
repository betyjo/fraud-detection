```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
credit = pd.read_csv(
    "../data/raw/creditcard.csv"
)

credit.head()
```

```python
credit.drop_duplicates(inplace=True)
```

```python
credit["Class"].value_counts()
```

```python
credit["Class"].value_counts(normalize=True)*100
```

```python
sns.countplot(
    x="Class",
    data=credit
)
```

Fraud transactions represent less than 1% of observations.
Strong imbalance exists.

```python
sns.histplot(
    credit["Amount"],
    bins=100
)
```

```python
sns.boxplot(
    x="Class",
    y="Amount",
    data=credit
)
```

```python
corr = credit.corr()
```

```python
plt.figure(figsize=(16,12))

sns.heatmap(
    corr,
    cmap="coolwarm"
)

plt.show()
```

```python
corr["Class"].sort_values()
```

```python
important_features = [
    "V4",
    "V10",
    "V12",
    "V14",
    "V17"
]

for col in important_features:

    plt.figure(figsize=(8,4))

    sns.boxplot(
        x="Class",
        y=col,
        data=credit
    )

    plt.show()
```
